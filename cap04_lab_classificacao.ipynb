## 4.7.2 Regressão logística

A primeira tarefa é ajustar uma regressão logística para prever `Direction` usando `Lag1` até `Lag5` e `Volume`.

A função `sm.GLM()` ajusta modelos lineares generalizados. Para obter uma regressão logística, usamos `family=sm.families.Binomial()`.

```python
allvars = Smarket.columns.drop(['Today', 'Direction', 'Year'])
design = MS(allvars)
X = design.fit_transform(Smarket)
y = Smarket.Direction == 'Up'
glm = sm.GLM(y,
             X,
             family=sm.families.Binomial())
results = glm.fit()
summarize(results)
```

O menor valor-p aparece em `Lag1`, com coeficiente negativo. A interpretação inicial é que, se o mercado teve retorno positivo no dia anterior, a probabilidade estimada de subida no dia atual tende a diminuir. Ainda assim, o valor-p é alto, então não há evidência estatística forte dessa associação no modelo completo.

Podemos acessar diretamente os coeficientes estimados com `params`.

```python
results.params
```

Os valores-p podem ser acessados com `pvalues`.

```python
results.pvalues
```

O método `predict()` retorna probabilidades estimadas. Sem passar uma nova matriz de desenho, ele calcula as probabilidades para os próprios dados de treino.

```python
probs = results.predict()
probs[:10]
```

Para transformar probabilidades em classes, usamos um limiar de 0,5. Probabilidades maiores que 0,5 são rotuladas como `Up`; as demais como `Down`.

```python
labels = np.array(['Down']*1250)
labels[probs > 0.5] = "Up"
```

A matriz de confusão compara as classes previstas com as classes reais. A diagonal indica acertos e os elementos fora da diagonal indicam erros.

```python
confusion_table(labels, Smarket.Direction)
```

Calculamos a acurácia de treino. Aqui o modelo acertou aproximadamente 52,2% das observações, mas esse número é otimista porque o modelo foi avaliado nos mesmos dados usados para ajustá-lo.

```python
(507+145)/1250, np.mean(labels == Smarket.Direction)
```

Para uma avaliação mais realista, separamos os anos de 2001 a 2004 como treino e o ano de 2005 como teste. Esse desenho respeita a estrutura temporal do problema: treinar no passado e testar no futuro.

```python
train = (Smarket.Year < 2005)
Smarket_train = Smarket.loc[train]
Smarket_test = Smarket.loc[~train]
Smarket_test.shape
```

Agora ajustamos o modelo apenas nos dados anteriores a 2005 e fazemos previsões para 2005.

```python
X_train, X_test = X.loc[train], X.loc[~train]
y_train, y_test = y.loc[train], y.loc[~train]
glm_train = sm.GLM(y_train,
                   X_train,
                   family=sm.families.Binomial())
results = glm_train.fit()
probs = results.predict(exog=X_test)
```

Guardamos os rótulos reais de treino e teste.

```python
D = Smarket.Direction
L_train, L_test = D.loc[train], D.loc[~train]
```

Aplicamos o limiar de 0,5 para classificar as observações de 2005.

```python
labels = np.array(['Down']*252)
labels[probs > 0.5] = 'Up'
confusion_table(labels, L_test)
```

A acurácia de teste fica em torno de 48%, com erro perto de 52%. Esse resultado é pior que uma regra ingênua de chute aleatório equilibrado, sugerindo que o modelo completo não generaliza bem.

```python
np.mean(labels == L_test), np.mean(labels != L_test)
```

Como os valores-p do modelo completo foram fracos, testamos um modelo mais simples com apenas `Lag1` e `Lag2`. A ideia é reduzir variáveis pouco informativas, pois elas podem aumentar variância sem reduzir viés.

```python
model = MS(['Lag1', 'Lag2']).fit(Smarket)
X = model.transform(Smarket)
X_train, X_test = X.loc[train], X.loc[~train]
glm_train = sm.GLM(y_train,
                   X_train,
                   family=sm.families.Binomial())
results = glm_train.fit()
probs = results.predict(exog=X_test)
labels = np.array(['Down']*252)
labels[probs > 0.5] = 'Up'
confusion_table(labels, L_test)
```

Calculamos a acurácia total e a acurácia condicional nos dias em que o modelo prevê `Up`.

```python
(35+106)/252, 106/(106+76)
```

O modelo simples acerta cerca de 56% dos movimentos diários. Contudo, prever `Up` todos os dias também teria desempenho semelhante nesse teste. A parte mais útil está na acurácia condicional: quando o modelo prevê alta, ele acerta perto de 58% das vezes.

Agora estimamos probabilidades para dois novos cenários hipotéticos de `Lag1` e `Lag2`.

```python
newdata = pd.DataFrame({'Lag1':[1.2, 1.5],
                        'Lag2':[1.1, -0.8]});
newX = model.transform(newdata)
results.predict(newX)
```

## 4.7.3 Análise Discriminante Linear: LDA

A LDA também será usada para prever `Direction`, com treino antes de 2005 e teste em 2005. O estimador da LDA adiciona intercepto automaticamente, então removemos a coluna `intercept` da matriz de desenho.

```python
lda = LDA(store_covariance=True)
```

```python
X_train, X_test = [M.drop(columns=['intercept'])
                   for M in [X_train, X_test]]
lda.fit(X_train, L_train)
```

Após ajustar o modelo, `means_` mostra as médias dos preditores em cada classe. Essas médias são usadas para estimar os vetores de média de cada grupo.

```python
lda.means_
```

As classes ficam armazenadas em `classes_`. Isso permite saber a ordem em que probabilidades e médias são apresentadas.

```python
lda.classes_
```

As probabilidades a priori estimadas para cada classe ficam em `priors_`. Aqui elas indicam a proporção de dias `Down` e `Up` no conjunto de treino.

```python
lda.priors_
```

Os vetores discriminantes lineares ficam em `scalings_`. Eles indicam a combinação linear dos preditores usada na regra de decisão da LDA.

```python
lda.scalings_
```

Geramos previsões para o conjunto de teste.

```python
lda_pred = lda.predict(X_test)
```

A matriz de confusão da LDA fica praticamente igual à da regressão logística simples.

```python
confusion_table(lda_pred, L_test)
```

Também podemos obter probabilidades posteriores para cada classe com `predict_proba()`. Aplicar o limiar de 0,5 à probabilidade posterior da classe `Up` reproduz as previsões do classificador.

```python
lda_prob = lda.predict_proba(X_test)
np.all(
    np.where(lda_prob[:,1] >= 0.5, 'Up','Down') == lda_pred
)
```

Em problemas com mais de duas classes, a previsão é dada pela classe com maior probabilidade posterior. O código abaixo confirma essa lógica no caso binário.

```python
np.all(
    [lda.classes_[i] for i in np.argmax(lda_prob, 1)] == lda_pred
)
```

Podemos mudar o limiar de decisão. Por exemplo, para prever `Down` apenas quando a probabilidade posterior de queda for maior que 90%, usamos a primeira coluna de `lda_prob`, pois `classes_` mostrou que `Down` é a primeira classe.

```python
np.sum(lda_prob[:,0] > 0.9)
```

Nenhum dia de 2005 atingiu esse nível de certeza para a classe `Down`. Isso mostra que o classificador não produziu probabilidades muito extremas nesse conjunto de teste.

Os classificadores do `sklearn` seguem um padrão comum: primeiro instanciamos o modelo, depois ajustamos com `fit()` e, por fim, prevemos com `predict()`. Esse padrão facilita reuso, validação cruzada e comparação entre métodos.

## 4.7.4 Análise Discriminante Quadrática: QDA

A QDA é semelhante à LDA na interface, mas permite matrizes de covariância diferentes por classe. Isso torna a fronteira de decisão quadrática, e não linear.

```python
qda = QDA(store_covariance=True)
qda.fit(X_train, L_train)
```

Como na LDA, podemos acessar as médias por classe e as probabilidades a priori.

```python
qda.means_, qda.priors_
```

A QDA estima uma matriz de covariância para cada classe. Abaixo está a covariância estimada da primeira classe.

```python
qda.covariance_[0]
```

Como a regra da QDA é quadrática, não há um único conjunto de coeficientes lineares como na LDA. A previsão, porém, segue o mesmo padrão com `predict()`.

```python
qda_pred = qda.predict(X_test)
confusion_table(qda_pred, L_test)
```

Calculamos a acurácia da QDA no teste.

```python
np.mean(qda_pred == L_test)
```

Neste recorte, a QDA alcança acurácia próxima de 60%, melhor que regressão logística, LDA e KNN nos testes anteriores. Ainda assim, em dados financeiros, esse resultado exigiria validação mais robusta antes de qualquer uso prático.

```python
knn3 = KNeighborsClassifier(n_neighbors=3)
knn3_pred = knn3.fit(X_train, L_train).predict(X_test)
np.mean(knn3_pred == L_test)
```

Agora aplicamos KNN ao conjunto `Caravan`, em que a variável resposta `Purchase` indica se um indivíduo comprou ou não uma apólice de seguro de caravana.

```python
Caravan = load_data('Caravan')
Purchase = Caravan.Purchase
Purchase.value_counts()
```

Calculamos a proporção de compradores.

```python
348 / 5822
```

As variáveis explicativas são todas as colunas, exceto `Purchase`.

```python
feature_df = Caravan.drop(columns=['Purchase'])
```

Como o KNN depende de distância, a escala das variáveis importa. Padronizamos as colunas para média zero e desvio padrão um, evitando que variáveis em escala maior dominem a distância.

```python
scaler = StandardScaler(with_mean=True,
                        with_std=True,
                        copy=True)
```

Primeiro ajustamos o transformador aos dados, depois aplicamos a transformação.

```python
scaler.fit(feature_df)
X_std = scaler.transform(feature_df)
```

Reconstruímos um `DataFrame` com as variáveis padronizadas.

```python
feature_std = pd.DataFrame(
    X_std,
    columns=feature_df.columns);
feature_std.std()
```

Os desvios aparecem muito próximos de 1. Pequenas diferenças decorrem da convenção usada no cálculo de variância: algumas funções usam divisor `n`, outras usam `n-1`.

Agora separamos treino e teste, deixando 1.000 observações para teste. O `random_state=0` garante reprodutibilidade.

```python
(X_train,
 X_test,
 y_train,
 y_test) = train_test_split(feature_std,
                            Purchase,
                            test_size=1000,
                            random_state=0)
```

No Jupyter, `?train_test_split` abre a documentação da função.

```python
?train_test_split
```

Ajustamos KNN com `K=1` no conjunto de treino e avaliamos no teste.

```python
knn1 = KNeighborsClassifier(n_neighbors=1)
knn1_pred = knn1.fit(X_train, y_train).predict(X_test)
np.mean(y_test != knn1_pred), np.mean(y_test != "No")
```

A taxa de erro do KNN fica em torno de 11%. Esse número não é necessariamente bom, pois uma regra que prevê sempre `No` já teria erro perto de 6,7%, dado o forte desbalanceamento da classe.

Nesse problema, a métrica mais interessante pode ser a precisão entre os indivíduos previstos como compradores.

```python
confusion_table(knn1_pred, y_test)
```

Entre os indivíduos previstos como `Yes`, calculamos a proporção que realmente comprou.

```python
9/(53+9)
```

### Parâmetro de ajuste: número de vizinhos

O número de vizinhos `K` é um hiperparâmetro. Testamos valores de 1 a 5 e avaliamos a precisão entre os indivíduos classificados como compradores.

```python
for K in range(1,6):
    knn = KNeighborsClassifier(n_neighbors=K)
    knn_pred = knn.fit(X_train, y_train).predict(X_test)
    C = confusion_table(knn_pred, y_test)
    templ = ('K={0:d}: # predicted to rent: {1:>2},' +
             ' # who did rent {2:d}, accuracy {3:.1%}')
    pred = C.loc['Yes'].sum()
    did_rent = C.loc['Yes','Yes']
    print(templ.format(
        K,
        pred,
        did_rent,
        did_rent / pred))
```

### Comparação com regressão logística

Ajustamos uma regressão logística com `sklearn`. O parâmetro `C` controla regularização; ao usar um valor muito alto, aproximamos a regressão logística não regularizada.

```python
logit = LogisticRegression(C=1e10, solver='liblinear')
logit.fit(X_train, y_train)
logit_pred = logit.predict_proba(X_test)
logit_labels = np.where(logit_pred[:,1] > 0.5, 'Yes', 'No')
confusion_table(logit_labels, y_test)
```

Com limiar 0,5, nenhuma observação do teste é classificada como compra. Em dados desbalanceados, o limiar padrão pode ser inadequado.

Ao reduzir o limiar para 0,25, o modelo passa a prever alguns compradores.

```python
logit_labels = np.where(logit_pred[:,1] > 0.25, 'Yes', 'No')
confusion_table(logit_labels, y_test)
```

Calculamos a precisão entre os indivíduos previstos como compradores.

```python
9/(20+9)
```

## 4.7.7 Regressão linear e regressão de Poisson nos dados `Bikeshare`

Agora usamos o conjunto `Bikeshare`, que registra o número de aluguéis de bicicleta por hora em Washington, DC, entre 2010 e 2012. A variável resposta é `bikers`.

```python
Bike = load_data('Bikeshare')
```

Verificamos dimensões e nomes das variáveis.

### Regressão linear

Ajustamos primeiro uma regressão linear para prever `bikers` usando mês, hora, indicador de dia útil, temperatura e condição climática.

```python
X = MS(['mnth',
        'hr',
        'workingday',
        'temp',
        'weathersit']).fit_transform(Bike)
Y = Bike['bikers']
M_lm = sm.OLS(Y, X).fit()
summarize(M_lm)
```

Como `mnth` e `hr` são variáveis categóricas, um nível de cada uma funciona como referência. No ajuste acima, `Jan` e `hr[0]` são níveis de base; os demais coeficientes são interpretados em relação a essas categorias.

A seguir usamos uma codificação alternativa por soma, em que os coeficientes de cada variável categórica somam zero. Nesse esquema, os efeitos são interpretados como desvios em relação à média.

```python
hr_encode = contrast('hr', 'sum')
mnth_encode = contrast('mnth', 'sum')
```

```python
X2 = MS([mnth_encode,
         hr_encode,
         'workingday',
         'temp',
         'weathersit']).fit_transform(Bike)
M2_lm = sm.OLS(Y, X2).fit()
S2 = summarize(M2_lm)
S2
```

A escolha da codificação muda a interpretação dos coeficientes, mas não muda as previsões do modelo. Verificamos isso comparando os valores ajustados.

```python
np.sum((M_lm.fittedvalues - M2_lm.fittedvalues)**2)
```

```python
np.allclose(M_lm.fittedvalues, M2_lm.fittedvalues)
```

Para reproduzir o gráfico dos efeitos por mês, extraímos os coeficientes de `mnth`. Como a codificação por soma omite um nível, calculamos o coeficiente de dezembro como o negativo da soma dos demais meses.

```python
coef_month = S2[S2.index.str.contains('mnth')]['coef']
coef_month
```

```python
months = Bike['mnth'].dtype.categories
coef_month = pd.concat([
    coef_month,
    pd.Series([-coef_month.sum()],
              index=['mnth[Dec]'
             ])
])
coef_month
```

Plotamos os coeficientes mensais.

```python
fig_month, ax_month = subplots(figsize=(8,8))
x_month = np.arange(coef_month.shape[0])
ax_month.plot(x_month, coef_month, marker='o', ms=10)
ax_month.set_xticks(x_month)
ax_month.set_xticklabels([l[5] for l in coef_month.index], fontsize=20)
ax_month.set_xlabel('Month', fontsize=20)
ax_month.set_ylabel('Coefficient', fontsize=20);
```

Fazemos processo semelhante para os coeficientes de hora.

```python
coef_hr = S2[S2.index.str.contains('hr')]['coef']
coef_hr = coef_hr.reindex(['hr[{0}]'.format(h) for h in range(23)])
coef_hr = pd.concat([coef_hr,
                     pd.Series([-coef_hr.sum()], index=['hr[23]'])
                    ])
```

```python
fig_hr, ax_hr = subplots(figsize=(8,8))
x_hr = np.arange(coef_hr.shape[0])
ax_hr.plot(x_hr, coef_hr, marker='o', ms=10)
ax_hr.set_xticks(x_hr[::2])
ax_hr.set_xticklabels(range(24)[::2], fontsize=20)
ax_hr.set_xlabel('Hour', fontsize=20)
ax_hr.set_ylabel('Coefficient', fontsize=20);
```

### Regressão de Poisson

Agora ajustamos um modelo de Poisson para a contagem `bikers`. A estrutura muda pouco: usamos `sm.GLM()` com `family=sm.families.Poisson()`.

```python
M_pois = sm.GLM(Y, X2, family=sm.families.Poisson()).fit()
```

Extraímos os coeficientes associados a mês e hora, completando os níveis omitidos da codificação por soma.

```python
S_pois = summarize(M_pois)
coef_month = S_pois[S_pois.index.str.contains('mnth')]['coef']
coef_month = pd.concat([coef_month,
                        pd.Series([-coef_month.sum()],
                                  index=['mnth[Dec]'])])
coef_hr = S_pois[S_pois.index.str.contains('hr')]['coef']
coef_hr = pd.concat([coef_hr,
                     pd.Series([-coef_hr.sum()],
                               index=['hr[23]'])])
```

Reproduzimos os gráficos dos coeficientes do modelo de Poisson.

```python
fig_pois, (ax_month, ax_hr) = subplots(1, 2, figsize=(16,8))
ax_month.plot(x_month, coef_month, marker='o', ms=10)
ax_month.set_xticks(x_month)
ax_month.set_xticklabels([l[5] for l in coef_month.index], fontsize=20)
ax_month.set_xlabel('Month', fontsize=20)
ax_month.set_ylabel('Coefficient', fontsize=20)
ax_hr.plot(x_hr, coef_hr, marker='o', ms=10)
ax_hr.set_xticks(x_hr[::2])
ax_hr.set_xticklabels(range(24)[::2], fontsize=20)
ax_hr.set_xlabel('Hour', fontsize=20)
ax_hr.set_ylabel('Coefficient', fontsize=20);
```

Por fim, comparamos os valores ajustados pela regressão linear e pela regressão de Poisson. Como o modelo de Poisson é adequado para contagens, suas previsões são não negativas.

```python
fig, ax = subplots(figsize=(8, 8))
ax.scatter(M2_lm.fittedvalues,
           M_pois.fittedvalues,
           s=20)
ax.set_xlabel('Linear Regression Fit', fontsize=20)
ax.set_ylabel('Poisson Regression Fit', fontsize=20)
ax.axline([0,0], c='black', linewidth=3,
          linestyle='--', slope=1);
```

O modelo de Poisson tende a ser mais coerente quando a resposta é uma contagem, pois respeita a restrição de não negatividade. A regressão linear pode produzir valores negativos, o que não faz sentido para número de aluguéis.

Neste laboratório, `sm.GLM()` foi usado com família binomial para regressão logística e com família de Poisson para modelagem de contagens. Outras famílias, como Gamma, podem ser usadas para outros tipos de variável resposta.
